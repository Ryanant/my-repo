```python
def process_segment(segment):
    segment = segment.replace("D/","Z/").replace("D ",("Z "))
    sides = re.findall(r'\((.*?)\)', segment)
    print(sides)
    sides = sides[0]    
    print(sides)
    lines = segment.replace("("+sides+")","").replace("/","").replace("AM","A").replace("DM","O").replace('ST','F').replace("WB","O").replace("Z","D").strip()
    print(lines)
    result = [f"{x}{y}" for x, y in product(lines, sides)] 
    print(result)
        
        
    return result
```

```python
def combine_segment(segments):
    result = []
    segments = segments.replace('GK','G (C)')
    segments = re.sub(r'\bDM\b(,?)', r'DM (C)\1', segments)
    for segment in segments.split(','):
        result.extend(process_segment(segment.strip()))
    return result
```

```python
def apply_combinations(df, column):

    df["Positions"] = df[column].apply(combine_segment)
    df['Positions'] = df['Positions'].apply(replace_values)

    return df
```

```python
def replace_values(lst):
    new_lst = []
    for item in lst:
        item = item.replace('O', 'DM')
        item = item.replace('GC', 'GK')
        item = item.replace('FC', 'ST')
        item = item.replace('A', 'AM')
        item = item.replace('DMR', 'WBR')
        item = item.replace('DML', 'WBL')
        item = item.replace('DMC', 'DM')


        new_lst.append(item)
    return new_lst
```

```python
df = apply_combinations(raw_table, 'Position')
```

```python
unique_values = set([item for sublist in df['Positions'] for item in sublist])
```

```python
for position in unique_values:
    df[position] = df.apply(lambda x: 1 if position in x["Positions"] else 0, axis=1)
```

```python
latest_export_date = df['Date'].max()
df['gk_rating'] = ((5 * (df['Agi']+df['Ref']+ df['Cmd']+df['Cnt'])) + (3 * (df['1v1']+df['Ant']+df['Kic']+df['Pos'])) +(1 * (df['Acc']+df['Aer']+df['Dec']+df['Fir']+df['Han']+df['Pas']+df['TRO']+df['Vis'])))/40
df['of_rating'] = (((df['Acc']+ df['Pac'])*5) +  ((df['Dri']+ df['Ant'] + df['Cnt']+ df['Agi']+df['Bal']+ df['Jum'])*3) + (df['Det']+ df['Wor'] + df['Sta']+ df['Nat.1']))/32
df['rating'] = df['gk_rating'].where(df['Positions'] == 'GK', df['of_rating'])
df['physical'] = df['Acc']+df['Pac'] + df['Jum']
earliest_ratings = df.loc[df.groupby('UID')['Date'].idxmin(), ['UID', 'rating','physical','CA','PA']].rename(columns={'rating': 'first_rating','physical': 'first_physical','CA': 'first_CA','PA': 'first_PA'})
df = df.merge(earliest_ratings, on='UID', how='left')

df['rating_growth'] = df['rating'] - df['first_rating']
df['physical_growth'] = df['physical'] - df['first_physical']
df['CA_growth'] = df['CA'] - df['first_CA']

df['age_at_export'] = df['Age'].apply(np.floor)
df['current_actual_age'] = df.groupby(['UID'])['age_at_export'].transform('last')
df['is_latest_export'] = df['Date'] == latest_export_date
df_max_date = df[df['Date'] == latest_export_date]
df_max_date['in_latest_export'] = True
df = pd.merge(df, df_max_date, how="left", on=["UID"],suffixes = [None,'_current'])
df=df.round(2)


```

```python
df['Date'] = pd.to_datetime(df['Date'])

july_first_dates = df[df['Date'].dt.month == 7][df['Date'].dt.day == 1]
latest_july_first = july_first_dates['Date'].max()
start_of_latest_season_df = df[(df['Date'] == latest_july_first) ]
start_of_latest_season_df = start_of_latest_season_df[['UID','age_at_export']]
df = df.merge(start_of_latest_season_df, on='UID', how='left',suffixes=('','_start_of_season'))
```

```python
columns_to_check = ['GK', 'DC', 'WBR', 'WBL', 'DM', 'AMC', 'ST']
df['Positions'] = df.apply(lambda row: [col for col in columns_to_check if row[col] == 1], axis=1)
df['Positions'] = df['Positions'].apply(lambda x: ', '.join(x))
df['Positions_filter'] = df['Positions']+', All'
df['All'] = 1
team_df = df[df['Club'].str.contains(current_save, case=False, na=False)]
team_df = team_df[team_df['is_latest_export']==True]
df = df[['Name','Club','current_actual_age','Positions','Positions_filter','GK','DC','WBR','DM','WBL','AMC','ST','All','rating','rating_growth','physical','physical_growth','Pac','Acc','Jum','age_at_export','is_latest_export','in_latest_export','age_at_export_start_of_season']]
```

```python


app = dash.Dash(__name__, external_stylesheets=[dbc.themes.BOOTSTRAP])

# Callback function to update the datatable
@app.callback(
    Output(component_id='my_squad', component_property='data'),
    [
        Input(component_id='age_range_slider', component_property='value'),
        Input(component_id='position_filter', component_property='value'),
        Input(component_id='is_latest_export_filter', component_property='value'),
        Input(component_id='club_filter', component_property='value')
    ]
)
def update_datatable(selected_age_range, selected_positions, is_latest_export, club_filter):
    # Make a copy of the global df to avoid modifying it directly
    df_filtering = df.copy()
    df_filtering['Positions_filter'] = df_filtering['Positions_filter'].str.split(', ')

    # Filter the DataFrame based on the selected positions
    if 'All' in selected_positions:
        filtered_df = df_filtering
    else:
        filtered_df = df_filtering[df_filtering['Positions_filter'].apply(
            lambda x: any(pos in x for pos in selected_positions))]

    # Convert list in 'Positions' column to comma-separated strings for display
    filtered_df['Positions_filter'] = filtered_df['Positions_filter'].apply(lambda x: ', '.join(x))

    # Filter by 'is_latest_export'
    filtered_df = filtered_df[filtered_df['is_latest_export'].isin(is_latest_export)]

    # Filter by age range
    filtered_df = filtered_df[
        (selected_age_range[0] <= filtered_df['age_at_export']) &
        (filtered_df['age_at_export'] <= selected_age_range[1])
    ]

    # Filter by club name
    if club_filter:  # Only filter if club_filter is not empty
        filtered_df = filtered_df[filtered_df['Club'].str.contains(club_filter, case=False, na=False)]


    # Return the filtered DataFrame as a list of dictionaries
    return filtered_df.to_dict('records')


# Define the layout of the app
app.layout = html.Div([
    dbc.Row(html.H1("Web App", style={'text-align': 'center'})),

    dbc.Row([
        dbc.Col(html.Div([
            dcc.RangeSlider(15, 40,
                            value=[15, 40],
                            id='age_range_slider'),
        ]), width=3),

        dbc.Col(html.Div([
            dcc.Dropdown(id="position_filter",
                         options=[{'label': pos, 'value': pos} for pos in ['GK', 'DC', 'WBR', 'WBL', 'DM', 'AMC', 'ST', 'All']],
                         multi=True,
                         value=['All']),  # Default value
        ]), width=3),

        dbc.Col(html.Div([
            dcc.Dropdown(id="is_latest_export_filter",
                         options=[{'label': 'True', 'value': True}, {'label': 'False', 'value': False}],
                         multi=True,
                         value=[True]),  # Default value
        ]), width=3),

        dbc.Col(html.Div([
            dcc.Input(id="club_filter",
                      type='text',
                      placeholder='Enter a club name...',
                      debounce=True,
                      value=""),  # Default value
        ]), width=3),
    ], className='row'),

    html.Br(),
    dbc.Container(
        dbc.Row(
            dbc.Col(
                html.Div(
                    id='output_container',
                    children=[
                        dash_table.DataTable(
                            columns=[{"name": i, "id": i} for i in df.columns],  # Include Positions column
                            data=df.to_dict('records'),
                            page_size=15,
                            id='my_squad',
                            filter_action="native",
                            sort_action="native",
                            sort_mode="multi",
                            row_selectable="multi",
                            hidden_columns=['is_latest_export', 'Positions_filter','GK','DC','WBR','WBL','DM','AMC','All','ST']  # Adjust hidden columns
                        )
                    ]
                ),
            ),
            className='row'
        ), fluid=True
    )
])


```

```python
import pandas as pd
from itertools import combinations

# Helper function to calculate the total rating of a combination
def calculate_combination_rating(combination):
    return sum([player['rating'] for player in combination])  # Access player directly, no need for player[1]

# Helper function to check for duplicates and return the first duplicate player
def find_duplicates(all_players):
    player_names = [player['Name'] for player in all_players]  # Accessing player name directly from Series
    seen = set()
    for player in all_players:
        if player['Name'] in seen:
            return player['Name']  # Return the first duplicate player
        seen.add(player['Name'])
    return None  # No duplicates found

# 1. Top 10 combinations of 6 DCs (Defender Centers)
dc_players = team_df[team_df['DC'] == 1]
dc_combinations = combinations(dc_players.iterrows(), 6)
dc_combination_ratings = [(comb, calculate_combination_rating([player[1] for player in comb])) for comb in dc_combinations]

# Sort DC combinations by rating and keep top 10
dc_combination_ratings.sort(key=lambda x: x[1], reverse=True)  # Sort by rating (highest first)
dc_combination_ratings = dc_combination_ratings[:20]  # Keep top 10

# 2. Top 10 combinations of 4 AMCs (Attacking Midfielders)
amc_players = team_df[team_df['AMC'] == 1]
amc_combinations = combinations(amc_players.iterrows(), 4)
amc_combination_ratings = [(comb, calculate_combination_rating([player[1] for player in comb])) for comb in amc_combinations]
amc_combination_ratings.sort(key=lambda x: x[1], reverse=True)
amc_combination_ratings = amc_combination_ratings[:15]

# 3. Top 10 combinations of 4 DMs (Defensive Midfielders)
dm_players = team_df[team_df['DM'] == 1]
dm_combinations = combinations(dm_players.iterrows(), 4)
dm_combination_ratings = [(comb, calculate_combination_rating([player[1] for player in comb])) for comb in dm_combinations]
dm_combination_ratings.sort(key=lambda x: x[1], reverse=True)
dm_combination_ratings = dm_combination_ratings[:15]

# 4. Top 5 combinations of 2 WBR (Wing-Back Right)
wbr_players = team_df[team_df['WBR'] == 1]
wbr_combinations = combinations(wbr_players.iterrows(), 2)
wbr_combination_ratings = [(comb, calculate_combination_rating([player[1] for player in comb])) for comb in wbr_combinations]
wbr_combination_ratings.sort(key=lambda x: x[1], reverse=True)
wbr_combination_ratings = wbr_combination_ratings[:10]

# 5. Top 5 combinations of 2 WBL (Wing-Back Left)
wbl_players = team_df[team_df['WBL'] == 1]
wbl_combinations = combinations(wbl_players.iterrows(), 2)
wbl_combination_ratings = [(comb, calculate_combination_rating([player[1] for player in comb])) for comb in wbl_combinations]
wbl_combination_ratings.sort(key=lambda x: x[1], reverse=True)
wbl_combination_ratings = wbl_combination_ratings[:10]

# 6. Top 5 combinations of 2 ST (Strikers)
st_players = team_df[team_df['ST'] == 1]
st_combinations = combinations(st_players.iterrows(), 2)
st_combination_ratings = [(comb, calculate_combination_rating([player[1] for player in comb])) for comb in st_combinations]
st_combination_ratings.sort(key=lambda x: x[1], reverse=True)
st_combination_ratings = st_combination_ratings[:10]

# 7. Top 1 combination of 2 GK (Goalkeepers)
gk_players = team_df[team_df['GK'] == 1]
gk_combinations = combinations(gk_players.iterrows(), 2)
gk_combination_ratings = [(comb, calculate_combination_rating([player[1] for player in comb])) for comb in gk_combinations]
gk_combination_ratings.sort(key=lambda x: x[1], reverse=True)
gk_combination_ratings = gk_combination_ratings[:1]

# Now we have combinations for each position: DCs, WBRs, WBLs, DMs, AMCs, STs, and GKs.

# Initialize variables for best team
best_team_rating = 0
best_team_combination = []
combination_counter = 0  # Counter for total combinations checked
successful_combination_counter = 0  # Counter for total combinations checked

# Iterate through all combinations for each position
print("\nBeginning search for best team combination...")

# Iterate through all combinations for each position
for dc_comb in dc_combination_ratings:
    for wbr_comb in wbr_combination_ratings:
        for wbl_comb in wbl_combination_ratings:
            for dm_comb in dm_combination_ratings:
                for amc_comb in amc_combination_ratings:
                    for st_comb in st_combination_ratings:
                        for gk_comb in gk_combination_ratings:
                            # Combine all selected players, extracting the actual data (player[1])
                            all_players = [player[1] for player in dc_comb[0]] + \
                                          [player[1] for player in wbr_comb[0]] + \
                                          [player[1] for player in wbl_comb[0]] + \
                                          [player[1] for player in dm_comb[0]] + \
                                          [player[1] for player in amc_comb[0]] + \
                                          [player[1] for player in st_comb[0]] + \
                                          [player[1] for player in gk_comb[0]]
                            
                            # Check for duplicates and print the duplicate player if found
                            duplicate_player = find_duplicates(all_players)
                            if duplicate_player:
                                # Increment the combination counter
                                combination_counter += 1
                            
                                # Print update every 20 combinations
                                if combination_counter % 50000 == 0:
                                    print(f"Checked {combination_counter} combinations so far. Current Best Team Rating: {best_team_rating}")
                                continue  # Skip this combination if there's a duplicate player
                            
                            # Calculate total rating if no duplicates
                            total_rating = sum([player['rating'] for player in all_players])
                            
                            # Update the best team if we found a better combination
                        if total_rating > best_team_rating:
                            best_team_rating = total_rating
                            best_team_combination = all_players
                            new_combination_counter += 1
                            print(f"Found {new_combination_counter} combinations so far. Current Best Team Rating: {best_team_rating}")

                            # Print the best team players by group and their ratings
                            print("\nNew Best Team Combination:")

                            # Initialize position categories with empty lists to store players by their specific combination
                            position_categories = {
                                "GK": [],
                                "DC": [],
                                "WBR": [],
                                "WBL": [],
                                "DM": [],
                                "AMC": [],
                                "ST": []
                            }

                            # Keep track of players and their positions directly from the combination
                            # For each combination, assign players to the corresponding position
                            for dc_comb in dc_comb[0]:
                                position_categories["DC"].append(dc_comb[1])  # Append the actual player to the DC position group

                            for wbr_comb in wbr_comb[0]:
                                position_categories["WBR"].append(wbr_comb[1])  # Append the actual player to the WBR position group

                            for wbl_comb in wbl_comb[0]:
                                position_categories["WBL"].append(wbl_comb[1])  # Append the actual player to the WBL position group

                            for dm_comb in dm_comb[0]:
                                position_categories["DM"].append(dm_comb[1])  # Append the actual player to the DM position group

                            for amc_comb in amc_comb[0]:
                                position_categories["AMC"].append(amc_comb[1])  # Append the actual player to the AMC position group

                            for st_comb in st_comb[0]:
                                position_categories["ST"].append(st_comb[1])  # Append the actual player to the ST position group

                            for gk_comb in gk_comb[0]:
                                position_categories["GK"].append(gk_comb[1])  # Append the actual player to the GK position group

                            # Display the best team organized by positions
                            for position, players in position_categories.items():
                                print(f"\n{position} Players:")
                                for player in players:
                                    print(f"{player['Name']} - Rating: {player['rating']}")

                            
                         

                            

# Output the best team combination
print(f"\nTotal Combinations Checked: {combination_counter}")
print("\nBest Team Combination (No Repeating Players):")

# Initialize position categories
position_categories = {
    "GK": [],
    "DC": [],
    "WBR": [],
    "WBL": [],
    "DM": [],
    "AMC": [],
    "ST": []
}

# Assign players to their actual positions in the best team
for player in best_team_combination:
    for position in ["GK", "DC", "WBR", "WBL", "DM", "AMC", "ST"]:
        if player[position] == 1:  # Check if the player can play the position
            position_categories[position].append(player)
            break  # Assign the first matching position and stop checking further

# Display the best team organized by positions
for position, players in position_categories.items():
    print(f"\n{position} Players:")
    for player in players:
        print(f"{player['Name']} - Rating: {player['rating']}")

print(f"\nTotal Rating of Best Team: {best_team_rating}")
```
