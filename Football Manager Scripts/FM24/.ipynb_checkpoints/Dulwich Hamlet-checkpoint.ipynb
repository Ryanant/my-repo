```python
python_exports = "C:/Users/jackj/Documents/Sports Interactive/Football Manager 2024/Python Exports"
current_save = "Havant - GS"
all_players_file = "all_players_data.csv"
folder_path = python_exports + "/" + current_save
all_players_file_path = folder_path+"/all_players/" + all_players_file
all_folder_files = os.listdir(folder_path)
all_csv_files = [f for f in all_folder_files if f.endswith('.csv')]    
all_csv_files = [w.replace('.csv', '') + ' 00:00:01' for w in all_csv_files]
```

```python
if not os.path.exists(all_players_file_path):
    exported_dates = pd.DataFrame({'Date': [1, 2]})

else:
    # If the file exists, read the existing data and make a copy    
    exported_dates = pd.read_csv(all_players_file_path)

exported_dates = exported_dates.Date.unique().tolist()
print(exported_dates)
```

```python
for file in all_csv_files:
    if file not in exported_dates:
        file_name = file.replace(' 00:00:01','')
        full_path = folder_path + '/' + file_name + '.csv'
        raw_table = pd.read_csv(full_path, sep=';', encoding='latin1')        
        date = file_name.split('.')[0] +  ' 00:00:01'
        raw_table['Date'] = pd.to_datetime(date, format = "%Y-%m-%d %H:%M:%S")
        raw_table['DoB'] = pd.to_datetime(raw_table['Date Of Birth'], dayfirst=True, errors='coerce')
        raw_table['Age'] = (raw_table['Date']-raw_table['DoB']).dt.days/365.25

        if not os.path.exists(all_players_file_path):
            raw_table.to_csv(all_players_file_path, index=False)
            print(f"Created file: {all_players_file_path}")
        else:
            # If the file exists, read the existing data and make a copy
            timestr = time.strftime("%Y%m%d-%H%M%S")
            existing_df = pd.read_csv(all_players_file_path)


            # Union the existing DataFrame with the new DataFrame
            combined_df = pd.concat([existing_df, raw_table]).drop_duplicates().reset_index(drop=True)

            # Save the combined DataFrame back to the CSV file
            combined_df.to_csv(all_players_file_path, index=False)
            print(f"Updated file: {all_players_file_path}")
```

```python
all_players_table = pd.read_csv(all_players_file_path)
all_players_table
```

```python
import re


# 2. Convert percentages to float
def convert_percent(val):
    if isinstance(val, str) and '%' in val:
        try:
            return round(float(val.replace('%', '')), 2)
        except ValueError:
            return val
    return val

all_players_table = all_players_table.applymap(convert_percent)

# 3. Round floats to 2 decimal places
all_players_table = all_players_table.applymap(lambda x: round(x, 2) if isinstance(x, float) else x)

# 4. Convert date columns to datetime
date_cols = [col for col in all_players_table.columns if 'date' in col or 'dob' in col]
for col in date_cols:
    all_players_table[col] = pd.to_datetime(all_players_table[col], errors='coerce')
```

```python
all_players_table['DoB'] = pd.to_datetime(all_players_table['DoB'], dayfirst=True, errors='coerce')
all_players_table['Date'] = pd.to_datetime(all_players_table['Date'], errors='coerce')


# Define a function to get Jan 1st of the football season
def get_season_jan1(date):
    if pd.isna(date):
        return pd.NaT
    return pd.Timestamp(year=date.year if date.month < 7 else date.year + 1, month=1, day=1)

# Create the 'Season Jan 1' date
all_players_table['Season Jan 1'] = all_players_table['Date'].apply(get_season_jan1)

# Calculate age on Jan 1 of the season
all_players_table['Age as of Jan 1'] = (
    all_players_table['Season Jan 1'].dt.year - all_players_table['DoB'].dt.year
    - all_players_table.apply(
        lambda row: (row['Season Jan 1'].month, row['Season Jan 1'].day) < (row['DoB'].month, row['DoB'].day),
        axis=1
    )
)

```

```python
latest_export_date = all_players_table['Date'].max()
earliest_ratings = all_players_table.loc[all_players_table.groupby('Unique ID')['Date'].idxmin(), ['Unique ID', 'GK Rating','FS Rating']].rename(columns={'GK Rating': 'First GK Rating','FS Rating': 'First FS Rating'})
all_players_table = all_players_table.merge(earliest_ratings, on='Unique ID', how='left')
all_players_table['GK Growth'] = all_players_table['GK Rating'] - all_players_table['First GK Rating']
all_players_table['FS Growth'] = all_players_table['FS Rating'] - all_players_table['First FS Rating']

all_players_table['is_latest_export'] = all_players_table['Date'] == latest_export_date
df_max_date = all_players_table[all_players_table['Date'] == latest_export_date]
df_max_date['in_latest_export'] = True
all_players_table = pd.merge(all_players_table, df_max_date, how="left", on=["Unique ID"],suffixes = [None,'_current'])
all_players_table=all_players_table.round(2)


```
